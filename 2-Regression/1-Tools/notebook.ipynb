```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets, linear_model, model_selection
```

```python
# Unpacking 2 element tuple returned by load_diabetes() function into X and y variables
X, y = datasets.load_diabetes(return_X_y=True)
print(X.shape)
print(X[0])
```

### What do X and y represent?

X (Features / Predictors): The indepentdent variables or input data fed into the model. In a 2D plot, this would correspond to the x-axis

y (Target / Outcome): The dependent variable or the target measurement that you want your model to predict. In a graph, this would correspond to the y-axis

```python
X = X[:, 2]
X = X.reshape((-1,1))
```

```python
# Splitting the dataset into training and testing sets
X_train, X_test, y_train, y_test = model_selection.train_test_split(X, y, test_size=0.33)
```

```python
# Training and fitting the datasets
model = linear_model.LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
plt.scatter(X_test, y_test,  color='black')
plt.plot(X_test, y_pred, color='blue', linewidth=3)
plt.xlabel('Scaled BMIs')
plt.ylabel('Disease Progression')
plt.title('A Graph Plot Showing Diabetes Progression Against BMI')
plt.show()
```

### Linear Regression Scatter Plot

Here we are comparing our linear regression model (the blue line) that was made using the training data and visually comparing it to the test data (the black dots).
